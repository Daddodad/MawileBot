```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline

# --- Model name ---
model_name = "LorenzoDeMattei/GePpeTto"

# --- Load tokenizer ---
tokenizer = AutoTokenizer.from_pretrained(model_name)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# --- Load base model (FP32) ---
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    device_map="auto"  # usa GPU se disponibile, altrimenti CPU
)
model.eval()

# --- Create pipeline ---
chatbot = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)
```

```python
import random

def clean_and_truncate(text, prompt):
    """
    text: stringa generata dal modello
    prompt: prompt originale (per rimuoverlo)
    """
    # 1️⃣ Rimuovo il prompt se presente
    if text.startswith(prompt):
        text = text[len(prompt):].strip()
    
    # 2️⃣ Trovo tutti i punti
    periods = [i for i, c in enumerate(text) if c in (".","?","!")]
    
    if periods:
        if len(periods) == 1:
            # Se c'è un solo punto, tronco lì
            cut_index = periods[0] + 1
        elif len(periods) >= 2:
            r = random.random()
            if r < 0.4:
                cut_index = periods[0] + 1        # 40% primo punto
            elif r < 0.6:
                cut_index = periods[1] + 1        # 20% secondo punto
            else:
                cut_index = random.choice(periods) + 1  # 40% uno a caso
        text = text[:cut_index].strip()
    
    return text

# --- Esempio di utilizzo ---
prompt = "<positivo> Oggi è stata una giornata lunga."
prompt = "Sei bravissimo"
prompt = 'Sei Scemo'
prompt = "Sableye sei bravissimo"
prompt = "dammi una risposta intelligente"


prompt = prompt + '.'

output_raw = chatbot(prompt,
                     max_new_tokens=50,
                     do_sample=True,
                    )[0]["generated_text"]
print(output_raw)
print('\n')
output_clean = clean_and_truncate(output_raw, prompt)
print(output_clean)
output_clean = clean_and_truncate(output_raw, prompt)
print(output_clean)
output_clean = clean_and_truncate(output_raw, prompt)
print(output_clean)
output_clean = clean_and_truncate(output_raw, prompt)
print(output_clean)
```
