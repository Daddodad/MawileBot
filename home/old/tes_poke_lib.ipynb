```python
# nome del file di input e output
input_file = "D:\Desktop_fake\MawileBot\home\old\Doc Evo.txt"
output_file = "ordinato.txt"

# leggi tutte le righe
with open(input_file, "r", encoding="utf-8") as f:
    lines = [line.strip() for line in f if line.strip()]  # rimuove righe vuote e spazi

# ordina alfabeticamente
lines.sort(key=str.lower)

# salva il risultato
with open(output_file, "w", encoding="utf-8") as f:
    for line in lines:
        f.write(line + "\n")

print("File ordinato salvato come", output_file)
```

```python
import pypokedex as poke
def esiste(name):
    try:
        poke.get(name=name)
        return True
    except:
        return False

def controlla_pokemon(file_path):
    with open(file_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            # prendi solo le parole (ignorando i numeri)
            parti = [x for x in line.split() if not x.isdigit()]
            for nome in parti:
                if not esiste(nome):  # la tua funzione
                    print(nome)

# usa il file ordinato
controlla_pokemon("ordinato.txt")
```
