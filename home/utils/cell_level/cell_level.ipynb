```python
def get_lvl_list(gym_cells):
    lvl = []
    gyms_passed = 0
    rows_passed = 0
    for i in range(0,22):
        lvl.append(5 + i + gyms_passed - rows_passed)
        if i+1 in gym_cells:
            gyms_passed += 1
        if (i+1)%7 == 0:
            rows_passed += 1
    gyms_passed = 0
    rows_passed = 0
    for i in range(1,21):
        lvl.append(lvl[21] + 2*i + gyms_passed - 2*rows_passed)
        if i+22 in gym_cells:
            gyms_passed += 1
        if (i+22)%7 == 0:
            rows_passed += 1
    return lvl
```

```python
gym_cell = [4,5,7,13,14,18,19,21,23,25,28,29,31,34,35,38,40,42]

print(get_lvl_list(gym_cell))
```
