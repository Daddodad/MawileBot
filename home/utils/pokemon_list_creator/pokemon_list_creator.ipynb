```python
import pypokedex as poke

choices = []
for i in range(0,1030):
    try:
        p =poke.get(dex = i)
        print(p.name)
        choices.append(p.name)
    except:
        print(i)
        pass
for i in range(10000,10300):
    try:
        p =poke.get(dex = i)
        print(p.name)
        choices.append(p.name)
    except:
        print(i)
        pass
```

```python
print(choices)
import json
with open("pokemon_list.json", "w", encoding="utf-8") as f:
    json.dump(choices, f, ensure_ascii=False, indent=2)
```

```python
# Over 500 bst, 'classici'

import pypokedex as poke

choices = []
for i in range(0,1030):
    try:
        p =poke.get(dex = i)
        print(p.name)
        bst = sum(poke.get(name=p.name).base_stats)
        if bst >= 500:
            choices.append(p.name)
    except:
        print(i)
        pass
```

```python
print(choices)
print(len(choices))
```
