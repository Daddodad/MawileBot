```python
import pypokedex as poke

choices = []
for i in range(0,1110):
    try:
        p =poke.get(dex = i)
        print(p.types)
        choices.append(sorted(p.types))
    except:
        print(i)
        pass
print(choices)

choices.append("")
```

```python
import json
from collections import Counter

type_list = choices

# Save the original list
with open('type_list.json', 'w') as f:
    json.dump(type_list, f, indent=2)

# Create frequency dictionary
type_tuples = [tuple(types) for types in type_list]
counter = Counter(type_tuples)

# Convert keys back to lists as strings for JSON
type_list_frequency = {str(list(k)): v for k, v in counter.items()}

# Save the frequency dictionary
with open('type_list_frequency.json', 'w') as f:
    json.dump(type_list_frequency, f, indent=2)
```
